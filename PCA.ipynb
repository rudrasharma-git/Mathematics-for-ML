```python
import numpy as np
X = np.random.randint(1, 10, size=(2, 11))
X = X.T
print(X)
import matplotlib.pyplot as plt
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel('X1')
plt.ylabel('X2')
#shifting the data to origin

X_meaned = X - np.mean(X, axis=0)
print(X_meaned.T)
plt.scatter(X_meaned[:, 0], X_meaned[:, 1])
plt.xlabel('X1')
plt.ylabel('X2')
plt.subplots(1, 2, figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1])
plt.subplot(1, 2, 2)
plt.scatter(X_meaned[:, 0], X_meaned[:, 1])
plt.show()
plt.scatter(X[:, 0], X[:, 1])
plt.scatter(X_meaned[:,0],X_meaned[:,1])
C_mat = np.cov(X_meaned, rowvar=False)
C_mat
eval,evec = np.linalg.eig(C_mat)
print("Eigenvalues: ", eval)
print("Eigenvectors: ", evec)
sorted_indices = np.argsort(eval)[::-1]
sorted_eval = eval[sorted_indices]
print("Sorted Eigenvalues: ", sorted_eval)
sorted_evec = evec[:, sorted_indices]
print("Sorted Eigenvectors: ", sorted_evec)
n=1
evec_subset = sorted_evec[:, 0:n]
print("Eigenvector subset: ", evec_subset)

X_reduced = np.dot(X_meaned, evec_subset)
print("Reduced data: ", X_reduced)
from sklearn.decomposition import PCA
pca = PCA(n_components = 1)
pca.fit(X_meaned)
X_res = pca.transform(X_meaned)
print("Reduced data using sklearn PCA: ", X_res)
```
